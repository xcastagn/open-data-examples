#### BARPA-R (BARPA/output-CMIP6/DD/AUS-15)
BARPA-R is a moderate-resolution (17 km) regional climate projections over Australia and the surrounding regions including parts of southeast Asia and New Zealand. This data set includes hourly, 3-hourly, daily and monthly data for temperature, moisture, wind and flux variables at sub-surface, surface, and pressure levels, and heights above surface. The vertical levels include 1000, 950, 925, 850, 700, 600, 500, 400, 300, 250, 200, 150, 100, 70, 50, 30, 20, 10 hPa pressure levels (18) and 10, 50, 100, 150, 200, 250 and 1500 metre above surface.

#### BARPA-R Convective Parameters (BARPA/output-CMIP6/DD/AUST-15)
BARPA-R is a moderate-resolution (17 km) regional climate projections over Australia and the surrounding regions including parts of southeast Asia and New Zealand.
This dataset includes 3-hourly data for a collection of convective parameters - diagnostics used to characterise environments favourable for thunderstorms and their associated hazards (heavy rainfall, large hail, damaging wind gusts, and tornadoes) - over Australia. These include: parcel-based indices such as convective available potential energy (CAPE) and convective inhibition (CIN) for surface-based, mixed-layer (ML), and most-unstable (MU) parcels; thermodynamic parameters such as bulk wind difference (BWD); and composite parameters such as derecho composite (DCP), significant tornado parameter (STP). Daily and monthly mean and maximum values are also available for surface-based CAPE and CIN.

#### BARPA-C - Convective-Scale (BARPA/output-CMIP6/DD/AUST-04)
BARPA-C is a kilometre-scale (4.4 km) regional climate projection, over Australia.
The data set includes 20-minute, hourly, 3-hourly, daily and monthly data for temperature, moisture, wind and flux variables at sub-surface, surface, and pressure levels , and heights above surface. There are additional variables unique to convection-permitting models, namely number of lightning flash, updraft helicity, column-maximum updraft, and derived radar reflectivity. The vertical levels include 1000, 975, 950, 925, 900, 850, 800, 750, 700, 600, 500, 400, 300, 250, 200, 150, 100, 70, 50, 30, 20, 10 hPa pressure levels (22) and 50, 100, 150, 200, 250 and 1500 metre above surface.

#### CORDEX-CMIP6-BARPA-R (BARPA/CORDEX-CMIP6/DD/AUS-20i)
BARPA-R is a moderate-resolution (17 km) regional climate projections over Australia and the surrounding regions including parts of southeast Asia and New Zealand - CORDEX-Australasia region.
The model is registered with source_id BARPA-R1-NN.
A subset of BARPA-R modelling variables are regridded for CMIP6-CORDEX standard grid AUS-20i. These variables belong to the CMIP6-CORDEX ATMOS CORE variables. They include hourly, daily and monthly data for surface and near-surface variables for pressure, temperature, moisture, wind, precipitation and radiation.

#### Setting up

First, let's define a function to load BARPA data into memory

````python
# Load required Python modules 

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import xarray as xr
import sys, os
import glob
import cartopy.crs as ccrs
from  matplotlib import cm
````

````python
# Lists of Global Climate Models that have been downscaled so far

model_list_barpar = [
    "ACCESS-CM2",
    "ACCESS-ESM1-5",
    "CESM2",
    "CMCC-ESM2",
    "EC-Earth3",
    "MPI-ESM1-2-HR",
    "NorESM2-MM",
]

model_list_barpac = [
    "ACCESS-ESM1-5",
    "EC-Earth3",
]
````

````python
# function to load in model data. Will change on AWS Open Data

def load_barpa_data(domain="*",gcm='*',scenario='*',rev='*',rcm='*',freq='*',var='*',year='*',month1='*',month2='*'):

    """
    Load a BARPA projection dataset for a specified domain, gcm, scenario, 
    revision, rcm, variable, frequency, and year–month period.

    This function constructs the expected file path for a BARRA2 NetCDF file, 
    then opens the dataset with xarray. It assumes the
    file exists and follows the standard BARPA2 directory and naming
    conventions.

    Wildcards can be used to load multiple years or months of data at once or
    when the filepath of a single model run is adequately identified with a
    subset of identifiers. However, an error will be raised if multiple runs
    are specified to be loaded at once

    Parameters
    ----------
    model_id : str
        Identifier for the model configuration (e.g., "BARRA-C2", "BARRA-R2").
    domain_id : str
        Spatial domain identifier used in the BARRA2 directory structure.
    freq : str
        Temporal frequency of the dataset (e.g., "1hr", "day").
    variable : str
        Variable name (e.g., "tas", "pr") used in both the directory and filename.
    yearmonth : str
        Year–month string in YYYYMM format specifying the target file period.

    Returns
    -------
    xarray.Dataset
        The opened BARRA2 dataset corresponding to the requested parameters.

    Notes
    -----
    This function does not perform error checking beyond the number of models read.
    If the constructed file path does not exist or the file is unreadable,
    `xr.open_dataset` will raise an exception.
    """
    filepath = f"/g/data/py18/BARPA/output-CMIP6/DD/{domain}/BOM/{gcm}/{scenario}/{rev}/{rcm}/v1-r1/{freq}/{var}/latest/"
    if freq == 'fx':
        filename = f"{var}_{domain}_{gcm}_{scenario}_{rev}_BOM_{rcm}_v1-r1_{freq}.nc"
    else:
        filename = f"{var}_{domain}_{gcm}_{scenario}_{rev}_BOM_{rcm}_v1-r1_{freq}_{year}{month1}-{year}{month2}.nc"
    paths = glob.glob(filepath)
    assert len(paths)>0, f"Requested filepath {filepath} does not exist"
    assert len(paths)==1, f"Multiple model runs matched at {filepath}"

    ds = xr.open_mfdataset(os.path.join(filepath,filename))
    
    return ds
````

#### Example 1: local warming 

Climate stripes (#ShowYourStripes) are a useful visualisation of temperature trends, created to intuitively compare trend signals to internal variability noise in annual temperature timeseries. They were developed by Ed Hawkins at the University of Reading in 2018. 

In this exercise we are plotting the climate stripes for Sydney(or your alternative location)  according to the BARPA-R ensemble of downscaled CMIP6 models under the scenario labelled SSP370. 


````python
# open data files
dict_da_tas_model = {}
for model in model_list_barpar:
    print(model)
    # Load historical data
    metadata =  {'domain':'AUS-15',
                       'gcm':model,
                       'scenario':'historical',
                       'rev':'*',
                       'year':'*',
                       'month1':'01',
                       'rcm':'BARPA-R',
                       'freq':'mon',
                       'month2':'12',
                     'var':'tas'}
    ds_tas_hist = load_barpa_data(**metadata)
    # Load ssp data
    metadata['scenario']='ssp370'
    ds_tas_ssp = load_barpa_data(**metadata)
    
    dict_da_tas_model[model] = xr.concat(
        [ds_tas_hist, ds_tas_ssp],
        dim="time"
    )["tas"]
````

````python
#  subset and load into memory
dict_da_tas_model_mn = {}
for model in model_list_barpar:
    print(model)
    dict_da_tas_model_mn[model] = dict_da_tas_model[model].sel(lon=slice(x0-0.5,x0+0.5),
                                                               lat=slice(y0-0.5,y0+0.5))
    dict_da_tas_model_mn[model] = dict_da_tas_model_mn[model].mean(["lat", "lon"])
    dict_da_tas_model_mn[model] = dict_da_tas_model_mn[model].resample({"time":"1Y"}).mean().compute()
````

````python
#Because some BARPA models use different calendars we overwrite the time coordinate with the year integer.
for model in model_list_barpar:
    dict_da_tas_model_mn[model] = dict_da_tas_model_mn[model].assign_coords(
        time=dict_da_tas_model_mn[model].time.dt.year
    )
````

````python
# merge into a single data array
da_tas_model = xr.concat(
    [dict_da_tas_model_mn[model] for model in model_list_barpar],
    pd.Index(model_list_barpar, name="model")
)
````

````python
# normalise by historical period
da_tas_model_normed = da_tas_model - da_tas_model.sel(time=slice(1985,2014)).mean('time')

# set up colours for figure
norm = plt.Normalize(-da_tas_model_normed.max(), da_tas_model_normed.max())
cmap = plt.cm.get_cmap('RdBu_r')
colors = cmap(norm(da_tas_model))

# plot climate stripes
for i,model in enumerate(model_list_barpar):
    plt.subplot(7,1,i+1)
    tmp = da_tas_model_normed.isel(model=[i])
    tmp['model'] = [0]
    plt.pcolormesh(tmp.values,cmap=cmap,norm=norm)
    plt.xticks([])
    plt.yticks([])
````

#### Example 2: extreme hourly rainfall: BARPA-R
Short duration high intensity rainfall is projected to increase in a warmer climate due to the increased capacity of a warmer atmosphere to hold moisture. 

This example calculates the maximum rain rate in a 100x100 km box around Sydney for each year between 1960 and 2100 according to the 7 BARPA ensemble members. Each data-point represents the average rainfall across a BARPA-R grid-box, which is a 17x17 km box. Although the timeseries are noisy, a clear increasing trend is present. 


````python
# read data files

# we use the variable prhmax, which contains the maximum hourly rainrate in each day. 
dict_da_prhmax_model = {}
for model in model_list_barpar:
    # Load historical data
    metadata =  {'domain':'AUS-15',
                       'gcm':model,
                       'scenario':'historical',
                       'rev':'*',
                       'year':'*',
                       'month1':'01',
                       'rcm':'BARPA-R',
                       'freq':'day',
                       'month2':'12',
                     'var':'prhmax'}
    ds_prhmax_hist = load_barpa_data(**metadata)
    # Load ssp data
    metadata['scenario']='ssp370'
    ds_prhmax_ssp = load_barpa_data(**metadata)
    
    dict_da_prhmax_model[model] = xr.concat(
        [ds_prhmax_hist, ds_prhmax_ssp],
        dim="time"
    )["prhmax"]
````

````python
# subset and load into memory
dict_da_prhmax_model_mn = {}
for model in model_list_barpar:
    print(model)
    dict_da_prhmax_model_mn[model] = dict_da_prhmax_model[model].sel(lon=slice(x0-0.5,x0+0.5),
                                                                     lat=slice(y0-0.5,y0+0.5))
    dict_da_prhmax_model_mn[model] = dict_da_prhmax_model_mn[model].max(["lat", "lon"])
    dict_da_prhmax_model_mn[model] = dict_da_prhmax_model_mn[model].resample({"time":"1Y"})
    dict_da_prhmax_model_mn[model] = dict_da_prhmax_model_mn[model].max().compute()
````

````python
# plot timeseries of annual maximums
for model in model_list_barpar:
    plt.plot(dict_da_prhmax_model_mn[model]["time"].dt.year, 3600*dict_da_prhmax_model_mn[model], label=model,lw=1)
    plt.legend()
    plt.title("Annual maximum hourly precipitation rate")
````

#### Example 4: Tropical Cyclone Visualisation

BARPA simulates realistic, alternate reality weather events. These can be used to 'war-game' weather events in a warmer climate. 

As shown in example 2, moist thermodynamics imply that the atmosphere will be able to carry more moisture in a warmer climate. This means that rainfall events have the potential to become more intense in the future. While CMIP6 models are able to simulate the increasing levels of atmospheric moisture, downscaled projections are needed to see the effect on small-scale weather systems that global models do not resolve.

This case-study looks at BARPA-R downscaling of EC-Earth3. Of all models downscaled with BARPA-R, EC-Earth3 shows the largest increase in atmospheric moisture.

We have identified two tropical low events from BARPA-R-EC-Earth3 with similar tracks paths and minimum pressures. The first event was simulated in an environment representing past climate, and the second future climate. The tutorial demonstrates how the rainfall and atmospheric moisture vary between the two lows.

This example plots the vertically intgrated atmospheric moisture vapour (purple colours) and hourly rainfall (bright colours) signal associated with the system over a three-day period. The magenta line indicates the track path of the central pressure.

It also compares Hovmöller diagrams of the longitude-maximum rain rates and vertically intgrated atmospheric moisture vapour for each system. 

````python
# Now, let's load our first tropical low, which occurred in the BARPA-R downscaling of EC-Earth in January 2009
# Remember this is a synthetic, model-generated track, it didn't occur in the real world.
# the variables we're loading are:
#   psl: sea level pressure
#   pr: precipitation
#   prw: vertically intgrated atmospheric moisture vapour

# set up metadata for loading function
metadata =  {'domain':'AUS-15',
                   'gcm':'EC-Earth3',
                   'scenario':'historical',
                   'rev':'r1i1p1f1',
                   'year':'2009',
                   'month1':'01',
                   'rcm':'BARPA-R',
                   'freq':'day',
                   'month2':'12'}


dict_hist_vars = {}

# dict_hist_vars # historical
# dict_proj_vars # projections
for variable_id in ['psl','pr','prw']:
    metadata['var']=variable_id    
    dict_hist_vars[variable_id] = load_barpa_data(**metadata)
    dict_hist_vars[variable_id] = dict_hist_vars[variable_id].sel(time=slice("2009-01-05","2009-01-16"))[variable_id].load()
    
# Now, let's change the precipitation units into something more understandable
dict_hist_vars['pr'][:] = dict_hist_vars['pr'][:]*3600*24
dict_hist_vars['pr'].attrs['units']='mm day-1'
````

````python
# Now, let's load our second tropical low, which occurred in the BARPA-R downscaling of EC-Earth in Decenver 2088 and January 2089
# Remember this is a synthetic, model-generated track
# the variables we're loading are:
#   psl: sea level pressure
#   pr: precipitation
#   prw: vertically intgrated atmospheric moisture vapour

# set up metadata for loading function
metadata =  {'domain':'AUS-15',
               'gcm':'EC-Earth3',
               'scenario':'ssp370',
               'rev':'r1i1p1f1',
               'month1':'01',
               'rcm':'BARPA-R',
               'freq':'day',
               'month2':'12'}

# Now we load the projected data. This system is split across two years

dict_proj_vars = {}
for variable_id in ['psl','pr','prw']:

    metadata['var']=variable_id
    metadata['year']='2088'    
    tmp1 =  load_barpa_data(**metadata)
    metadata['year']='2089'    
    tmp2 =  load_barpa_data(**metadata)
    dict_proj_vars[variable_id] = xr.concat([tmp1,tmp2],'time')
    dict_proj_vars[variable_id] = dict_proj_vars[variable_id].sel(time=slice("2088-12-25","2089-01-06"))[variable_id].load()

# Now, let's change the precipitation units into something more understandable
dict_proj_vars['pr'][:] = dict_proj_vars['pr'][:]*3600*24
dict_proj_vars['pr'].attrs['units']='mm day-1'
````

Now let's find those lows!

In this example, the systems we're interested in are the lowest-pressure objects in the box 100 E-150 E, 30S - 0S. So we can just find the minimum over space at each timestep to create the track paths

````python
track_longitude, track_latitude = [0,0],[0,0]
for i,da in enumerate([dict_hist_vars,dict_proj_vars]):
    
    # subset to a box surrounding the lows
    subsetted_pressure = da['psl'].sel(lon=slice(100,150),lat=slice(-40,0))
    
    # extract latitude and longitudes
    subsetted_lon = subsetted_pressure.lon
    subsetted_lat = subsetted_pressure.lat
    
    # find location of minimum pressure at each time
    track_xindex = subsetted_pressure.min('lat').argmin('lon')
    track_yindex = subsetted_pressure.min('lon').argmin('lat')
    
    # find coordinates of minimum pressures
    track_longitude[i] = subsetted_lon[track_xindex]
    track_latitude[i] = subsetted_lat[track_yindex]

````

````python
# Visualise 12 snapshots of the cyclone track

for i,da in enumerate([dict_hist_vars,dict_proj_vars]):
    fig = plt.figure(figsize=(12,8))
    for j in range(12):
        # set up axes with geometric projection
        ax = plt.subplot(3,4,j+1, projection=ccrs.PlateCarree()) 
        
        # plot hourly precip rate, with sec-> hour conversion rate from snapshot time
        pcolor1 = (da['prw'][j]).plot(vmin=0,
                                         vmax=100,
                                         cmap='Purples',
                                         add_colorbar=False) 
        
        # plot hourly precip rate, with sec-> hour conversion rate from snapshot time
        pcolor2 = (da['pr'][j].where(da['pr'][j]>10)).plot(vmin=0,
                                                             vmax=150,
                                                             cmap='turbo',
                                                             add_colorbar=False) 
        # draw coastlines
        ax.coastlines() 
        ax.set_xlim(105,160)
        ax.set_ylim(-45,-7)
        
        # set timestamp as axes label
        ax.set_title(da['prw'].time[j].values.__str__()[:16]) 
        
        # plot full cyclone track
        ax.plot(track_longitude[i][:-1],track_latitude[i][:-1],c='magenta') 
        
        # plot instantaneous cyclone location from snapshot time
        plt.plot(track_longitude[i][j],track_latitude[i][j],'o',c='magenta',ms=4,lw=1) 

    # add colour scales to figure
    fig.subplots_adjust(top=0.95, bottom=0.1)
    cax1 = fig.add_axes([0.1,0.05,0.3,0.03])
    fig.colorbar(pcolor1,
                 cax=cax1,
                 orientation='horizontal')
    cax1.set_xlabel("Atmospheric moisture vapour, kg/m2")
    cax2 = fig.add_axes([0.6,0.05,0.3,0.03])
    fig.colorbar(pcolor2,
                 cax=cax2,
                 orientation='horizontal')
    cax2.set_xlabel("Precipitation, mm/day")
    fig.suptitle(['Historical Cyclone','Future Cyclone'][i])
````

To summarise the differences between the cyclones, we can plot 'Hovmöller' diagrams, which collapse the 3D (lat, lon, time) data along one spatial dimension.

In this case, we take the maximum of rainfall in the longitude band between 100 and 150 East.

### Example 5: East Coast Lows in BARPA-R and BARPA-C

This final example contrasts the behaviour of a synthetic synoptic weather system in BARPA-R and BARPA-C.

The system is an East Coast Low, a category of intense low-pressure system that forms near the eastern coast of Australia, bringing heavy rain, strong winds, and large waves.

This system was simulated in May 2055 under conditions projected by ACCESSS-ESM1.5. Its formation on the 22nd of the month was similar in BARPA-C and BARPA-R. According to BARPA-R, it moves shoreward and de-intensifies on the 24th of May, whereas according to BARPA-C, it remains strong and moves southwards down the East Coast of Australia. 

This example plots the mean sea level pressure (black contours) and hourly rainfall (colours) signal associated with the system over a three-day period. The magenta line indicates the track path of the central pressure.

The rainfall from BARPA-C includes features which appear realistic, with squall lines, embedded thunderstorms and frontal structures. By contrast, the rainfall from BARPA-R is more blocky and unrealistic.

````python
# Now, let's load our East Coast Low in BARPA-R
# Remember this is a synthetic, model-generated track, it didn't occur in the real world!
# the variables we're loading are:
#   psl: sea level pressure
#   pr: precipitation

metadata =  {'domain':'AUS-15',
                   'gcm':'ACCESS-ESM1-5',
                   'scenario':'ssp370',
                   'rev':'r6i1p1f1',
                   'year':'2055',
                   'month1':'01',
                   'rcm':'BARPA-R',
                   'freq':'1hr',
                   'month2':'12'}

dict_barpar_vars = {}

# dict_hist_vars # historical
# dict_proj_vars # projections
for variable_id in ['psl','pr','prw']:
    metadata['var']=variable_id    
    dict_barpar_vars[variable_id] = load_barpa_data(**metadata)[variable_id]
    dict_barpar_vars[variable_id] = dict_barpar_vars[variable_id].sel(time=slice("2055-05-22","2055-05-26 00")).load()
    
# Now, let's change the precipitation units into something more understandable
dict_barpar_vars['pr'][:] = dict_barpar_vars['pr'][:]*3600
dict_barpar_vars['pr'].attrs['units']='mm hr-1'
````

````python
# Now, let's load the same East Coast Low in BARPA-C
metadata =  {'domain':'AUST-04',
                   'gcm':'ACCESS-ESM1-5',
                   'scenario':'ssp370',
                   'rev':'r6i1p1f1',
                   'year':'2055',
                   'month1':'05',
                   'rcm':'BARPA-C',
                   'freq':'1hr',
                   'month2':'05'}
dict_barpac_vars = {}
for variable_id in ['psl','pr','prw']:

    metadata['var']=variable_id
   
    dict_barpac_vars[variable_id] = load_barpa_data(**metadata)[variable_id]
    dict_barpac_vars[variable_id] = dict_barpac_vars[variable_id].sel(time=slice("2055-05-22","2055-05-26 00")).load()

# Now, let's change the precipitation units into something more understandable
dict_barpac_vars['pr'][:] = dict_barpac_vars['pr'][:]*3600
dict_barpac_vars['pr'].attrs['units']='mm hr-1'
````

````python
# We calculate the track path by looking for the minimum pressure in the box 150-160E, 40 - 15 S
track_longitude, track_latitude = [0,0],[0,0]
for i,da in enumerate([dict_barpac_vars,dict_barpar_vars]):
    
    # subset to a box surrounding the lows
    subsetted_pressure = da['psl'].sel(lon=slice(150,160),lat=slice(-40,-15))
    
    # extract latitude and longitudes
    subsetted_lon = subsetted_pressure.lon
    subsetted_lat = subsetted_pressure.lat
    
    # find location of minimum pressure at each time
    track_xindex = subsetted_pressure.min('lat').argmin('lon')
    track_yindex = subsetted_pressure.min('lon').argmin('lat')
    
    # find coordinates of minimum pressures
    track_longitude[i] = subsetted_lon[track_xindex]
    track_latitude[i] = subsetted_lat[track_yindex]
````

````python
# Visualise 12 snapshots of the ECL track
ax,pcolor1,pcolor2,dot = [0,0],[0,0],[0,0],[0,0]

for i,da in enumerate([dict_barpac_vars,dict_barpar_vars]):
    fig=plt.figure(figsize=(8,12)) # set up figure
    for j in range(12):
        # set up axes with geometric projection
        ax[i] = plt.subplot(3,4,j+1, projection=ccrs.PlateCarree()) 
        
        # plot sea level pressure from snapshot time
        pcolor1 = (da['psl'][6*j]/100).plot.contour(levels=np.arange(980,1025,5),colors='k',linewidths=[1])#vmax=101000,
                                         #vmin=98500,
                                         #cmap='Blues_r',
                                         #add_colorbar=False) 

        
        # plot hourly precip rate, with sec-> hour conversion rate from snapshop time
        pcolor2 = (da['pr'][6*j].where(da['pr'][6*j]>2)).plot(vmin=0,
                                                             vmax=50,
                                                             cmap='turbo',
                                                             add_colorbar=False) 
        # draw coastlines
        ax[i].coastlines() 
        ax[i].set_xlim(145,160)
        ax[i].set_ylim(-45,-15)
        
        # set timestamp as axes label
        ax[i].set_title(da['prw'].time[j*6].values.__str__()[:16]) 
        
        # plot full cyclone track
        ax[i].plot(track_longitude[i][:-3],track_latitude[i][:-3],c='magenta',lw=1) 
        
        # plot cyclone location at cyclone time
        dot[i] = plt.plot(track_longitude[i][6*j],track_latitude[i][6*j],'o',c='magenta',ms=6) 
    
    # add colour scales to figure        
    fig.subplots_adjust(top=0.95, bottom=0.2)
    cax2 = fig.add_axes([0.3,0.15,0.4,0.03])
    fig.colorbar(pcolor2,
                 cax=cax2,
                 orientation='horizontal')
    cax2.set_xlabel("Precipitation, mm/hr")

    fig.suptitle(['BARPA-C East Coast Low','BARPA-R East Coast Low'][i])

````
